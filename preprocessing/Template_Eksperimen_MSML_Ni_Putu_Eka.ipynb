```python
# Check data imbalance
# data jamur Edible (e) ada sedikit lebih banyak daripada data jamur Poisonous (p)

import seaborn as sns

balance = sns.histplot(data=mushroom_data, x="class")
for container in balance.containers:
    balance.bar_label(container)

label_map = {
  "p": "Poisonous (p)",
  "e": "Edible (e)"
}

ticks = balance.get_xticks()
old_labels = [label.get_text() for label in balance.get_xticklabels()]

balance.set_xticks(ticks)
balance.set_xticklabels([label_map.get(x, x) for x in old_labels]);
```

```python
# Normalisasi
# mengubah invalid value menjadi kategori "missing"
clean_data = mushroom_data.copy()
clean_data["stalk-root"] = clean_data["stalk-root"].replace("?", "missing")
```

```python
# drop kolom yang tidak perlu
clean_data = clean_data.drop("veil-type", axis=1)
clean_data
```

```python
from sklearn.preprocessing import OneHotEncoder
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split

label_encoder = LabelEncoder()

clean_data["class"] = label_encoder.fit_transform(
  clean_data["class"]
)

# Pisahkan fitur dan data
y = clean_data["class"]
X = clean_data.drop(columns=["class"])

# Encode kategori dengan One Hot Encoder karena nilai berupa kategori bukan rentang
encoder = OneHotEncoder(
  handle_unknown="ignore",
  sparse_output=False
)

X_encoded = encoder.fit_transform(X)

# Ubah ke dataframe
feature_names = encoder.get_feature_names_out(X.columns)

clean_data = pd.DataFrame(
  X_encoded,
  columns=feature_names
)

# Add encoded target
clean_data["class"] = y.to_numpy()
```

```python
clean_data.to_csv("processed_data.csv", index=False)
```
